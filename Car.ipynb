```python
#Transform petrrol,petorl,petrl to petrol
df["fuel"] = df["fuel"].replace({
    "Petrrol": "Petrol",
    "Petorl": "Petrol",
    "Petrl": "Petrol"
})
```

```python
#Count unique values in fuel
df["fuel"].value_counts()
```

```python
#Transform  Deisel,Diesl to Diesel
df["fuel"] = df["fuel"].replace({
    "Deisel": "Diesel",
    "Diesl": "Diesel"
})
```

```python
# Check fuel value counts after cleaning
df["fuel"].value_counts()
```

```python
# Transform Indvidual,Indivudal to Individual and transform Deelar ,Dealerr to Dealer
df["seller_type"] = df["seller_type"].replace({
    "Indvidual": "Individual",
    "Indivudal": "Individual",
    "Deelar": "Dealer",
    "Dealerr": "Dealer"
})
```

```python
# Check seller_type value counts after cleaning
df["seller_type"].value_counts()
```

```python
#Remove Outlier Years
df = df[df["year"] != 1800]
df = df[df["year"] != 3000]
```

```python
# Calculate car age 
df['car_age'] = 2026 - df['year']
print(df[["name","year", "car_age"]].head())
```

```python
# Extract brand name from the first word of car name
df['brand'] = (
    df['name'].dropna().astype(str).apply(lambda x: x.split()[0])
)
print(df[['name', 'brand']].head())
```

```python
#Calculate Kilometers Per Year
# Calculate annual average usage (km per year), using maximum to avoid division by zero
import numpy as np
df['km_per_year'] = df['km_driven'] / np.maximum(df['car_age'], 1)
print(df[['name', 'km_driven', 'car_age', 'km_per_year']].head())
```

```python
# Create price categories (Budget, Mid-Range, Luxury) based on selling price
def classify_price(price):
    if price < 250000:
        return 'Budget'
    elif price <= 600000:
        return 'Mid-Range'
    else:
        return 'Luxury'
df['price_category'] = df['selling_price'].apply(classify_price)
print(df[['name', 'selling_price', 'price_category']].head())
```

```python
# Selling Price Statistics
mean_price = df['selling_price'].mean()
print("Mean Price:", mean_price)

median_price = df['selling_price'].median()
print("Median Price:", median_price)

min_price = df['selling_price'].min()
print("Min Price:", min_price)

max_price = df['selling_price'].max()
print("Max Price:", max_price)

std_price = df['selling_price'].std()

print("Standard Deviation:", std_price)
```

```python
# Visualizing Selling Price Distribution
import matplotlib.pyplot as plt
plt.hist(df['selling_price'], bins=30)
plt.title('Selling Price Distribution')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
```

```python
# Statistics for car_age
mean_age = df['car_age'].mean()
print("Mean Age:", mean_age)

median_age = df['car_age'].median()
print("Median Age:", median_age)

std_age = df['car_age'].std()
print("Standard Deviation:", std_age)

min_age = df['car_age'].min()
print("Min Age:", min_age)

max_age = df['car_age'].max()
print("Max Age:", max_age)
```

```python

# Histogram for car_age
import matplotlib.pyplot as plt
plt.hist(df['car_age'], bins=20)
plt.title('Car Age Distribution')
plt.xlabel('Age (Years)')
plt.ylabel('Count')
plt.show()
```

```python
# Get the top 5 most common car brands in the dataset
top_brands = df['brand'].value_counts().head(5)
print(top_brands)
```

```python
#Plot bar chart for top 5 car brands
import matplotlib.pyplot as plt
plt.bar(top_brands.index, top_brands.values)
plt.title('Top 5 Car Brands')
plt.xlabel('Brand')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Scatter plot: Car Age vs Selling Price
plt.scatter(df['car_age'], df['selling_price'], alpha=0.5)
plt.title('Car Age vs Selling Price')
plt.xlabel('Car Age (Years)')
plt.ylabel('Selling Price')
plt.show()
```

```python
# Scatter plot comparing car age and price 
for fuel_type in df['fuel'].value_counts().head(3).index:
    subset = df[df['fuel'] == fuel_type]
    plt.scatter(subset['car_age'], subset['selling_price'], label=fuel_type, alpha=0.5)
plt.title('Car Age vs Selling Price by Fuel Type')
plt.xlabel('Car Age (Years)')
plt.ylabel('Selling Price')
plt.legend()
plt.show()
```
